```python
import os
import pickle
from deepface import DeepFace
```

```python
CLASS_ID = 1
DATASET_PATH = f"dataset/class_{CLASS_ID}"
ENCODINGS_DIR = "encodings"
OUTPUT_PKL_PATH = os.path.join(ENCODINGS_DIR, f"class_{CLASS_ID}_encodings.pkl")

# Create encodings directory if it doesn't exist
os.makedirs(ENCODINGS_DIR, exist_ok=True)

class_database = {}

print(f"Target Dataset Path: {DATASET_PATH}")
print(f"Output File will be: {OUTPUT_PKL_PATH}")
```

```python
# Verify that the folders have been renamed to Full Names
student_folders = os.listdir(DATASET_PATH)
print("Found the following student folders:")
print(student_folders)
```

```python
print(f"🚀 Starting Encoding Generation for Class {CLASS_ID}...")

# Loop through every folder ("Sahil Dhale", "Dhaval Dupare", etc.)
for student_name in student_folders:
    student_folder = os.path.join(DATASET_PATH, student_name)
    
    # Skip anything that isn't a folder (like hidden system files)
    if not os.path.isdir(student_folder):
        continue
        
    print(f"\n📸 Processing Student: {student_name}")
    student_embeddings = []

    # Read every photo in that student's folder
    for img_name in os.listdir(student_folder):
        img_path = os.path.join(student_folder, img_name)
        
        try:
            # ✅ The crucial RetinaFace + ArcFace + Alignment upgrade!
            result = DeepFace.represent(
                img_path=img_path, 
                model_name="Facenet", 
                detector_backend="retinaface", 
                enforce_detection=True, # Forces it to ignore bad photos without faces
                align=True  # 🔥 FIX: Added align=True to mathematically match the backend!
            )
            
            # Store the embedding
            student_embeddings.append(result[0]["embedding"])
            print(f"   ✅ Successfully encoded: {img_name}")
            
        except Exception as e:
            # If it's a landscape, a text file, or just a blurry photo, it safely skips it
            print(f"   ⚠️ Skipped {img_name} (No face found or bad image)")

    # Save the student's embeddings to the dictionary using their FULL NAME as the key
    if len(student_embeddings) > 0:
        class_database[student_name] = student_embeddings
    else:
        print(f"   ❌ WARNING: No valid faces found for {student_name}!")
```

```python
# Save the final dictionary to the .pkl file
with open(OUTPUT_PKL_PATH, "wb") as f:
    pickle.dump(class_database, f)

print(f"\n🎉 DONE! Encodings safely saved to: {OUTPUT_PKL_PATH}")
```
